```r
# Select key columns of interest based on the subsequent analyses
key_summary_cols <- c(
  "Average.Temperature..C",
  "total_emission",
  "Urban.population",
  "Rice.Cultivation",
  "On.farm.Electricity.Use",
  "Pesticides.Manufacturing",
  "Fertilizers.Manufacturing",
  "Food.Processing",
  "Manure.Management"
)

# Display summary statistics for these key variables
summary(data[, key_summary_cols])
```

```r
# Calculate and format summary statistics (including standard deviation) into a tidy table
summary_stats <- do.call(data.frame,
  lapply(data[, key_summary_cols], function(x) {
    c(Min = min(x, na.rm = TRUE),
      Q1 = quantile(x, 0.25, na.rm = TRUE),
      Median = median(x, na.rm = TRUE),
      Mean = mean(x, na.rm = TRUE),
      Q3 = quantile(x, 0.75, na.rm = TRUE),
      Max = max(x, na.rm = TRUE),
      `Std Dev` = sd(x, na.rm = TRUE),
      NAs = sum(is.na(x)))
  })
)

# Transpose and rename for better table formatting
summary_table <- t(summary_stats)
colnames(summary_table) <- c("Min", "1st Qu.", "Median", "Mean", "3rd Qu.", "Max", "Std Dev", "NAs")

# Output a highly professional table using knitr::kable
knitr::kable(summary_table, digits = 3, caption = "Key Variables: Descriptives & Spread Metrics")
```

```r
# 2. Heatmap of Correlations for a subset of major emission sources to keep it readable
library(reshape2)
library(ggplot2)

# Selecting some key variables based on the structure
key_cols <- c("total_emission", "Savanna.fires", "Forest.fires", "Crop.Residues",
              "Rice.Cultivation", "Food.Transport", "Forestland", "On.farm.energy.use")

# Filter numeric_data to key columns that exist
key_numeric_data <- numeric_data[, colnames(numeric_data) %in% key_cols]

# Calculate correlation
cor_matrix <- cor(key_numeric_data, use = "complete.obs")

# Melt the correlation matrix for ggplot
melted_cor <- melt(cor_matrix)

ggplot(data = melted_cor, aes(x=Var1, y=Var2, fill=value)) +
  geom_tile(color = "white") +
  scale_fill_gradient2(low = "blue", high = "red", mid = "white",
                       midpoint = 0, limit = c(-1,1), space = "Lab",
                       name="Correlation") +
  theme_minimal() +
  theme(axis.text.x = element_text(angle = 45, vjust = 1, hjust = 1)) +
  coord_fixed() +
  labs(title = "Correlation Heatmap of Key Emission Sources", x = "", y = "")
```

```r
# 1. Analysis for Response Variable: total_emission
# Define the predictors for emissions
emission_response <- "total_emission"
emission_predictors <- c("Pesticides.Manufacturing", "Fertilizers.Manufacturing", "Food.Processing", "Manure.Management")

# Subset and clean data beforehand to avoid passing na.action to graphical parameters
emission_data <- na.omit(data[, c(emission_response, emission_predictors)])

# Create the scatterplot matrix for total_emission vs its predictors
pairs(
  emission_data,
  main = "Scatterplot Matrix: Total Emissions & Predictors",
  pch = 19,
  col = "steelblue"
)

# Calculate and format the correlation matrix
emission_cor <- cor(emission_data, use = "complete.obs")
knitr::kable(emission_cor, digits = 3, caption = "Sample Correlations for Total Emissions")
```
